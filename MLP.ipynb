```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('mnist_784.csv')
np.random.seed(50)

# Activation function.
def sigmoid_activation(x):
    return 1 / (1 + np.exp(-x))

# Mean absolute error loss function.
def squared_loss_function(actual_value, predicted_value):
    return np.sum(np.square(actual_value - predicted_value))
    
    
# Derivative of the activation function.    
def sigmoid_activation_derivative(x):
    return x * (1-x)
    
def activation_calculation(weights_matrix, input_vector, bias_vector):
    dot_product_vector = (weights_matrix @ input_vector) + bias_vector
    activation_vector = sigmoid_activation(dot_product_vector)
    return activation_vector

# Hyper parameters.
neural_network = [784, 128, 10]
input_size = neural_network[0]
learning_rate = 0.01
epochs = 10

# Inputs.
X = (df.iloc[:, 0:input_size] / 255).values
# Labels
Y = df.iloc[:, input_size].values

# Training sets.
training_size = int(len(X) * 0.70)
X_training_set = X[:training_size]
Y_training_set = Y[:training_size]

# Test sets.
X_test_set = X[training_size:]
Y_test_set = Y[training_size:]

# Initialize neural network weights and biases (model parameters).
weights = []
biases = []

# Initialize weights with random values.
for current_in, next_out in zip(neural_network, neural_network[1:]):
    weights.append(np.random.randn(next_out, current_in)

# Initialize biases with random values.
for layer_size in neural_network[1:]:
    biases.append(np.random.randn(layer_size, 1))
    
# Training.
for epoch in range(epochs):
    
    error = 0
    
    for batch in range(10):
        X_batch_training_set=X_training_set[batch*1000:(batch+1)*1000]
        Y_batch_training_set=Y_training_set[batch*1000:(batch+1)*1000]
        
        # Forward pass over training data.
        for i in range(len(X_batch_training_set)):
            
            # Initialize input vector.
            input_vector = X_batch_training_set[i].reshape(input_size, 1)
        
            # Initialize label vector.
            actual_label_vector = np.zeros((10, 1))
            actual_label_vector[Y_batch_training_set[i]] = 1
            
            # Forward propagation.
            # Step 1. Start with the input_vector.
            activated_neuron = input_vector
            activated_neurons = []
            activated_neurons.append(activated_neuron)
    
            # Step 2. Iterate through the layers.
            for layer in range(len(neural_network)-1):
                activated_neuron = activation_calculation(weights[layer], activated_neuron, biases[layer])
                activated_neurons.append(activated_neuron)
                
            # Step 3. Last layer is the output.
            predicted_label_vector = activated_neurons[-1]
    
            # Calculating the error using the mean absolute error loss function.
            error += squared_loss_function(actual_label_vector, predicted_label_vector)
    
            # Backpropagation.
            # Step 1. Calculate the delta.
            delta = 2 * (activated_neuron - actual_label_vector) * sigmoid_activation_derivative(activated_neuron)
            
            # Step 2. Go backwards through layers.
            for layer in range(len(weights) - 1, -1, -1):
        
                # Step 3. Calculate partital derivative of loss function with respect to weight.
                weight_gradient = delta @ activated_neurons[layer].T
                
                # Step 4. Calculate partital derivative of loss function with respect to bias.
                bias_gradient = delta
                
                # Step 5. Calculate partital derivative of loss function with respect to the last neuron.
                previous_delta = (weights[layer].T @ delta) * sigmoid_activation_derivative(activated_neurons[layer])
                
                # Step 6. Delta is updated.
                delta = previous_delta
                
                # Step 7. Apply gradient descent to weight and bias. Go down the hill slowly.
                weights[layer] -= learning_rate * weight_gradient
                biases[layer] -= learning_rate * bias_gradient
        
        print(error)

# Testing the neural network.
correct_predictions = 0

for i in range(len(X_test_set)):
    
    # Initialize input vector.
    input_vector = X_test_set[i].reshape(input_size, 1)

    # Forward propagation.
    activated_neuron = input_vector

    # Iterate through the layers.
    for layer in range(len(neural_network)-1):
        activated_neuron = activation_calculation(weights[layer], activated_neuron, biases[layer])

    # Get the predicted digit.
    predicted_digit = np.argmax(activated_neuron)

    # Get the actual digit.
    actual_digit = Y_test_set[i]

    # Check if the prediction is correct.
    if predicted_digit == actual_digit:
        correct_predictions += 1

# Calculate accuracy.
accuracy = correct_predictions / len(X_test_set) * 100

print("Correct predictions:", correct_predictions)
print("Total test images:", len(X_test_set))
print(f"Test accuracy: {accuracy:.2f}%")
```
