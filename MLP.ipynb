```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('mnist_784.csv')
np.random.seed(50)

# Activation function.
def sigmoid_activation(x):
    return 1 / (1 + np.exp(-x))
    
# Derivative of the activation function.    
def sigmoid_activation_derivative(x):
    return x * (1-x)
    
def activation_calculation(weights_matrix, input_vector, bias_vector):
    dot_product_vector = (weights_matrix @ input_vector) + bias_vector
    activation_vector = sigmoid_activation(dot_product_vector)
    return activation_vector

# Hyper parameters.
neural_network = [784, 4, 10]
learning_rate = 0.001
epochs = 2

# Inputs.
X = (df.iloc[:, 0:neural_network[0]] / 255).values
# Labels
Y = df.iloc[:, neural_network[0]].values

# Training sets.
training_size = int(len(X) * 0.70)
X_training_set = X[:training_size]
Y_training_set = Y[:training_size]

# Test sets.
X_test_set = X[training_size:]
Y_test_set = Y[training_size:]

# Initialize neural network weights and biases (model parameters).
weights = []
biases = []

# Initialize weights with random values.
for current_in, next_out in zip(neural_network, neural_network[1:]):
    weights.append(np.random.randn(next_out, current_in))

# Initialize biases with random values.
for layer_size in neural_network[1:]:
    biases.append(np.random.randn(layer_size, 1))
    
# Training.
for epoch in range(epochs):
    
    error = 0
    
    # Forward pass over training data.
    for i in range(training_size):
        
        # Initialize input vector.
        input_vector = X[i].reshape(neural_network[0], 1)
    
        # Initialize label vector.
        actual_label_vector = np.zeros((10, 1))
        actual_label_vector[Y_training_set[i]] = 1
        
        # Forward propagation.
        # Step 1. Start with the input_vector.
        activated_neuron = input_vector
        activated_neurons = []
        activated_neurons.append(activated_neuron)

        # Step 2. Iterate through the layers.
        for layer in range(len(neural_network)-1):
            activated_neuron = activation_calculation(weights[layer], activated_neuron, biases[layer])
            activated_neurons.append(activated_neuron)
            
        # Step 3. Last layer is the output.
        predicted_label_vector = activated_neurons[-1]

        
        error += np.sum(np.square(np.absolute(actual_label_vector - predicted_label_vector)))

        # Backpropagation.
        # Step 1. Calculate the delta
        delta = (activated_neuron - actual_label_vector) * sigmoid_activation_derivative(activated_neuron)
        
        # Go backwards through layers
        for layer in range(len(weights) - 1, -1, -1):
    
            # Calculate gradients
            weight_gradient = delta @ activated_neurons[layer].T
            bias_gradient = delta
    
            # Calculate previous layer delta
            previous_delta = (weights[layer].T @ delta) * sigmoid_activation_derivative(activated_neurons[layer])
            delta = previous_delta
            
            # Gradient descent
            weights[layer] -= learning_rate * weight_gradient
            biases[layer] -= learning_rate * bias_gradient
    
    
    average_error = error / (2 * training_size)
    print(average_error)
```
