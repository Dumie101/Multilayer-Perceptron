```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('mnist_784.csv')
np.random.seed(50)

def sigmoid_activation(x):
    return 1 / (1 + np.exp(-x))
    
def sigmoid_activation_derivative(x):
    return x * (1-x)
    
def activation_calculation(weights_matrix, input_vector, bias_vector):
    dot_product_vector = (weights_matrix @ input_vector) + bias_vector
    activation_vector = sigmoid_activation(dot_product_vector)
    return activation_vector
    
# Network structure
structure = [784, 4, 4, 10]

num_input_neurons = 784

# Inputs
input_digit_vectors = (df.iloc[:, 0:num_input_neurons] / 255).values

# Correct labels
correct_input_classes = df.iloc[:, num_input_neurons].values

# Dataset sizes
training_size = int(len(input_digit_vectors) * 0.70)
test_size = int(len(input_digit_vectors) * 0.30)

# Initialize network
weights = []
biases = []

for current_in, next_out in zip(structure, structure[1:]):
    weights.append(np.random.randn(next_out, current_in) * 0.01)

for layer_size in structure[1:]:
    biases.append(np.zeros((layer_size, 1)))

learning_rate = 0.001

# Epochs
epochs = 20

for epoch in range(epochs):
    
    error = 0
    
    # Forward pass over training data
    for i in range(training_size):
        # Input
        input_digit_vector = input_digit_vectors[i].reshape(num_input_neurons, 1)
    
        # Expected output
        desired_input_vector = np.zeros((10, 1))
        desired_input_vector[correct_input_classes[i]] = 1
        
        # Forward propagation
        activated_neuron = input_digit_vector
        activated_neurons = []
        activated_neurons.append(input_digit_vector)
    
        for layer in range(len(weights)):
            activated_neuron = activation_calculation(weights[layer], activated_neuron, biases[layer])
            activated_neurons.append(activated_neuron)
            
        error += np.sum(np.square(np.absolute(desired_input_vector - activated_neuron)))
    
        delta = (activated_neuron - desired_input_vector) * sigmoid_activation_derivative(activated_neuron)
        
        # Go backwards through layers
        for layer in range(len(weights) - 1, -1, -1):
    
            # Calculate gradients
            weight_gradient = delta @ activated_neurons[layer].T
            bias_gradient = delta
    
            # Calculate previous layer delta
            previous_delta = (weights[layer].T @ delta) * sigmoid_activation_derivative(activated_neurons[layer])
            delta = previous_delta
            
            # Gradient descent
            weights[layer] -= learning_rate * weight_gradient
            biases[layer] -= learning_rate * bias_gradient
    
    
    average_error = error / (2 * training_size)
    print(average_error)
```
